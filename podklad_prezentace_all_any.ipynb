```python
# Naměřené teploty ze senzoru
teploty = [35, 42, 48, 51, 67, 45, 39]

# Kontrola, jestli jsou všechna měření v povoleném rozsahu
vsechna_v_poradku = all(
    0 <= teplota <= 80 for teplota in teploty
)

# Kontrola, jestli se objevila nebezpečně vysoká teplota
vysoka_teplota = any(
    teplota > 60 for teplota in teploty
)

# Výpočet průměrné teploty
prumer = sum(teploty) / len(teploty)

# Vyhodnocení výsledků
print("Všechna měření v pořádku:", vsechna_v_poradku)
print("Byla překročena bezpečná teplota:", vysoka_teplota)
print("Průměrná teplota:", round(prumer, 2), "°C")

if not vsechna_v_poradku:
    print("CHYBA: Naměřená hodnota je mimo rozsah!")

elif vysoka_teplota:
    print("VAROVÁNÍ: Teplota je příliš vysoká!")

else:
    print("Stroj pracuje v pořádku.")
```

```python
# Naměřené teploty ze senzoru
teploty = [35, 42, 48, 51, 67, 45, 39]

# Kontrolní proměnné
vsechna_v_poradku = True
vysoka_teplota = False

# Procházení všech naměřených teplot
for teplota in teploty:

    if teplota < 0 or teplota > 80:
        vsechna_v_poradku = False

    if teplota > 60:
        vysoka_teplota = True

# Výpočet průměrné teploty
prumer = sum(teploty) / len(teploty)

# Výpis výsledků
print("Všechna měření v pořádku:", vsechna_v_poradku)
print("Byla překročena bezpečná teplota:", vysoka_teplota)
print("Průměrná teplota:", round(prumer, 2), "°C")

# Vyhodnocení
if not vsechna_v_poradku:
    print("CHYBA: Naměřená hodnota je mimo rozsah!")

elif vysoka_teplota:
    print("VAROVÁNÍ: Teplota je příliš vysoká!")

else:
    print("Stroj pracuje v pořádku.")
```
